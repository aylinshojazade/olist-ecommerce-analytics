# Schema Validation — Olist E-Commerce Dataset

This notebook validates the structure and key constraints of the Olist dataset before setting up the relational database.

The main focus is on:
- identifying candidate primary keys
- validating uniqueness
- checking null values
- understanding identifiers and relationships

## 1. Customers

### 1.1 Primary Key Validation — `customer_id`

#### Question

Can `customer_id` serve as the primary key of the `customers` table?

To answer this, we check:
- total number of records
- number of unique `customer_id` values
- number of null `customer_id` values

```python
import pandas as pd

customers_df = pd.read_csv("../data/olist_customers_dataset.csv")

customers_df.head(10)
```

```python
customers_df.shape[0]
```

```python
customers_df["customer_id"].nunique()
```

```python
customers_df["customer_id"].isna().sum()
```

#### Insight

`customer_id` contains 99,441 unique and non-null values across 99,441 records. Therefore, it satisfies the uniqueness and non-null requirements for a primary key in the `customers` table.

### 1.2 Customer Identifier Validation — `customer_unique_id`

#### Question

Is `customer_unique_id` unique for each record in the `customers` table, or can the same customer appear in multiple records?

To answer this, we compare:
- total number of records
- number of unique `customer_unique_id` values
- number of null `customer_unique_id` values

```python
customers_df.shape[0]
```

```python
customers_df["customer_unique_id"].nunique()
```

```python
customers_df["customer_unique_id"].isna().sum()
```

#### Insight

`customer_id` is unique for every record, while `customer_unique_id` is not unique: there are 99,441 records but only 96,096 unique `customer_unique_id` values.

This suggests that `customer_id` identifies a customer record, whereas `customer_unique_id` represents a customer at the business level and may appear in multiple records.

The relationship between repeated `customer_unique_id` values and multiple orders will be validated later using the `orders` table.

## 2. Orders

```python
orders_df = pd.read_csv("../data/olist_orders_dataset.csv")
```

### 2.1 Primary Key Validation — `order_id`

#### Question

Can `order_id` serve as the primary key of the `orders` table?

To answer this, we check:
- total number of records
- number of unique `order_id` values
- number of null `order_id` values

```python
orders_df.head(10)
```

```python
orders_df.shape[0]
```

```python
orders_df["order_id"].nunique()
```

```python
orders_df["order_id"].isna().sum()
```

### Insight

`order_id` contains 99,441 unique and non-null values across 99,441
records. Therefore, it satisfies the uniqueness and non-null requirements
for a primary key in the `orders` table.

### 2.2 Foreign Key Validation — `customer_id`

#### Question

Does `orders.customer_id` correctly reference `customers.customer_id`?

To answer this, we need to check:
- whether `orders.customer_id` contains null values
- whether every `customer_id` in `orders` exists in the `customers` table

```python
orders_df["customer_id"].isna().sum()
```

```python
(~orders_df["customer_id"].isin(customers_df["customer_id"])).sum()
```

#### Insight

All `customer_id` values in `orders` exist in `customers`, so `customer_id` is a valid foreign key.

## 3. Order Items

```python
order_items_df = pd.read_csv("../data/olist_order_items_dataset.csv")
```

```python
order_items_df.head(10)
```

### 3.1 Primary Key Validation - `order_id`

#### Question

Can `order_id` alone serve as the primary key of the `order_items` table?

To answer this, we compare:

- total number of records
- number of unique `order_items.order_id` values
- number of null `order_items.order_id` values

```python
order_items_df.shape[0]
```

```python
order_items_df["order_id"].nunique()
```

```python
order_items_df["order_id"].isna().sum()
```

So `order_id` alone cannot be the primary key, because multiple rows belong to the same order.

### 3.2 Composite Primary Key Validation — `order_id`, `order_item_id`

#### Question

Can the combination of `order_id` and `order_item_id` uniquely identify each row in the `order_items` table?

To answer this, we need to check whether each combination of `order_id` and `order_item_id` is unique and non-null.

```python
order_items_df[["order_id", "order_item_id"]].shape[0]
```

```python
order_items_df[["order_id", "order_item_id"]].drop_duplicates().shape[0]
```

```python
order_items_df[["order_id", "order_item_id"]].isna().sum()
```

#### Insight

The combination of `order_id` and `order_item_id` is unique and non-null across all 112,650 records. Therefore, `(order_id, order_item_id)` can serve as the composite primary key of `order_items`.

### 3.3 Foreign Key Validation - `order_id`

#### Question 
Does every `order_id` in `order_items` exist in the `orders` table?

```python
(~order_items_df["order_id"].isin(orders_df["order_id"])).sum()
```

```python
order_items_df["order_id"].isna().sum()
```

All `order_id` values in `order_items` exist in the `orders` table, and none are null. Therefore, `order_items.order_id` satisfies the requirements for a foreign key.
