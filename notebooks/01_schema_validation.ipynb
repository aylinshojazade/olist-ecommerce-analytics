# Schema Validation — Customers

## 1. Primary Key Validation

### Question
Is `customer_id` a valid primary key?

We need to check:
- total number of rows
- number of unique `customer_id` values
- number of null `customer_id` values

```python
import pandas as pd

df = pd.read_csv("../data/olist_customers_dataset.csv")

df.head(10)
```

```python
df.shape[0]
```

```python
df["customer_id"].nunique()
```

```python
df["customer_id"].isna().sum()
```

### Insight

`customer_id` contains 99,441 unique and non-null values across 99,441
records. Therefore, it satisfies the uniqueness and non-null requirements
for a primary key in the `customers` table.

### Question

Is `customer_unique_id` a unique identifier for each customer in the customers table, or can the same customer appear in multiple records?

we need to check: 
- total number of rows
- number of unique `customer_unique_id` values
- number of null `customer_unique_id` values

```python
df.shape[0]
```

```python
df["customer_unique_id"].nunique()
```

```python
df["customer_unique_id"].isna().sum()
```

### Insight

`customer_id` is unique for every record, while `customer_unique_id` is
not unique: there are 99,441 records but only 96,096 unique
`customer_unique_id` values.

This suggests that `customer_id` identifies a customer record, whereas
`customer_unique_id` represents a customer at the business level and may
appear in multiple records.

The relationship between repeated `customer_unique_id` values and
multiple orders will be validated later using the `orders` table.
